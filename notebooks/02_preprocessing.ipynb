# Smart MCQ Solver Challenge

## Text Preprocessing

This notebook prepares the dataset for classical ML models
(TF-IDF) as well as Transformer models (BERT, DeBERTa).

```python
import pandas as pd
import numpy as np
import re
```

#Load the Dataset

```python
train = pd.read_csv("/content/train.csv")
test = pd.read_csv("/content/test.csv")
```

#Inspect one Example

```python
sample = train.iloc[0]

print(sample["prompt"])
print("\nA :", sample["A"])
print("B :", sample["B"])
print("C :", sample["C"])
print("D :", sample["D"])
print("E :", sample["E"])
print("\nCorrect :", sample["answer"])
```

#Function

```python
def clean_text(text):
    text = str(text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # Remove leading/trailing spaces
    text = text.strip()

    return text
```

#Testing Function

```python
example = "   This      is      a      sample.     "
print(clean_text(example))
```

#Cleaning


```python
text_columns = [
    "prompt",
    "A",
    "B",
    "C",
    "D",
    "E"
]

for column in text_columns:
    train[column] = train[column].apply(clean_text)
    test[column] = test[column].apply(clean_text)
```
