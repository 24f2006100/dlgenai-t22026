# Data Formatting for MCQ Classification

This notebook converts the original MCQ dataset into a format suitable
for machine learning and transformer models.

```python
import pandas as pd
```

```python
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

#Function

```python
def create_mcq_pairs(df):
    rows = []
    option_columns = ["A", "B", "C", "D", "E"]
    for _, row in df.iterrows():
        for option in option_columns:
            rows.append({
                "id": row["id"],
                "prompt": row["prompt"],
                "option": row[option],
                "option_id": option,
                "label": int(option == row["answer"])
            })
    return pd.DataFrame(rows)
```

```python
train_pairs = create_mcq_pairs(train)
```
